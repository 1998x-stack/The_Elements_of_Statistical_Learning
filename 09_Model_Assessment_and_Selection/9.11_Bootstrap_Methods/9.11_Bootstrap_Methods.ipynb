# 9.11 自举 (Bootstrap)（深讲）

用有效回抽样的重采样估计统计量的分布。

## 1. 重采样

从 N 个样本有放回抽 N 个构成一个重样本。

## 2. 估计标准误

对回归系数重复拟合并画分布直方图。

```
import numpy as np
from sklearn.linear_model import LinearRegression
rng=np.random.default_rng(0)
n=100
X=rng.normal(size=(n,3))
y=X@np.array([1.0,-2.0,0.5])+0.4*rng.standard_normal(n)
betas=[]
for _ in range(500):
    idx=rng.integers(0,n,n)
    betas.append(LinearRegression().fit(X[idx],y[idx]).coef_)
betas=np.array(betas)
print('自举标准误:', np.round(betas.std(axis=0),3))
```

```
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif']=['Arial Unicode MS','PingFang SC','SimHei']
plt.hist(betas[:,0],bins=40)
plt.xlabel('beta_1 自举分布'); plt.title('Bootstrap 分布')
plt.show()
```
