# 7.2 分段多项式与样条（深讲）

用截断幂基构造三次样条，并与全局高次多项式比较。

## 1. 截断幂基

$$(x-\xi)_+^3=\max(0,(x-\xi))^3$$ 在节点处引入二阶光滑的连接。

## 2. 拟合曲线

比较样条与 30 次多项式的行为。

```
import numpy as np
rng=np.random.default_rng(0)
x=np.sort(rng.uniform(0,1,120))
y=np.sin(4*x)+0.15*rng.standard_normal(120)
knots=[0.4,0.7]
def trunc(X,k): return np.maximum(X-k,0)**3
def basis(X):
    X=np.asarray(X)[:,None]
    B=[np.ones_like(X[:,0]),X[:,0],X[:,0]**2,X[:,0]**3]
    for k in knots: B.append(trunc(X[:,0],k))
    return np.column_stack(B)
print('基矩阵形状', basis(x).shape)
```

```
from sklearn.linear_model import LinearRegression
m=LinearRegression().fit(basis(x),y)
xx=np.linspace(0,1,200)
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif']=['Arial Unicode MS','PingFang SC','SimHei']
plt.scatter(x,y,s=8,alpha=0.4)
plt.plot(xx,m.predict(basis(xx)),label='三次样条')
plt.legend(); plt.title('样条基拟合')
plt.show()
```
