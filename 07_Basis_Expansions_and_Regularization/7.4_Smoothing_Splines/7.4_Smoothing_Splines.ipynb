# 7.4 平滑样条（深讲）

展示罚函数 $\lambda\int (f'')^2$ 如何控制平滑度。

## 1. 目标

$$\min_f \sum (y_i-f(x_i))^2+\lambda\int (f'')^2$$

## 2. 光滑度

不同 $\lambda$（/s）给出不同偏差-方差取舍。

```
import numpy as np
from scipy.interpolate import UnivariateSpline
rng=np.random.default_rng(0)
x=np.sort(rng.uniform(0,1,100)); y=np.sin(6*x)+0.2*rng.standard_normal(100)
xx=np.linspace(0,1,300)
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif']=['Arial Unicode MS','PingFang SC','SimHei']
plt.scatter(x,y,s=6,alpha=0.4)
for s in (0.1,0.6): plt.plot(xx, UnivariateSpline(x,y,s=s)(xx), label=f's={s}')
plt.plot(xx,np.sin(6*xx),'k--',label='真实')
plt.legend(); plt.title('平滑样条')
plt.show()
```
