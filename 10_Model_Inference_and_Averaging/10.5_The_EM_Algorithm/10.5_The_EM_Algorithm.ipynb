# 10.5 EM 算法（深讲）

手写高斯混合的 EM，观察对数似然单调递增。

## 1. E 步与 M 步

E: 计算责任; M: 用责任加权更新均值/方差/混合比例。

```
import numpy as np
from scipy.stats import norm
rng=np.random.default_rng(0)
x=np.concatenate([rng.normal(0,1,200),rng.normal(4,1.5,200)])
m1,m2,s1,s2,pi=1.0,3.0,1.0,1.0,0.5
def loglik(x,m1,m2,s1,s2,pi):
    return np.sum(np.log(pi*norm.pdf(x,m1,s1)+(1-pi)*norm.pdf(x,m2,s2)+1e-12))
print('iter  loglik')
for it in range(30):
    p1=pi*norm.pdf(x,m1,s1); p2=(1-pi)*norm.pdf(x,m2,s2); g=p1/(p1+p2+1e-12)
    m1=(g*x).sum()/g.sum(); m2=((1-g)*x).sum()/(1-g).sum()
    s1=np.sqrt((g*(x-m1)**2).sum()/g.sum()); s2=np.sqrt(((1-g)*(x-m2)**2).sum()/(1-g).sum())
    pi=g.mean()
    if it%5==0: print(it, round(loglik(x,m1,m2,s1,s2,pi),3))
```

```
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif']=['Arial Unicode MS','PingFang SC','SimHei']
gx=np.linspace(-4,9,300)
fit=pi*norm.pdf(gx,m1,s1)+(1-pi)*norm.pdf(gx,m2,s2)
plt.hist(x,bins=40,density=True,alpha=0.4)
plt.plot(gx,fit,'r')
plt.title('EM 拟合的高斯混合密度')
plt.show()
```
