# 10.7 Bagging（深讲）

对重采样后的决策树做预测平均以降低方差。

## 1. 单棵树 vs Bagging

比较测试 MSE。

## 2. 收敛

随 B 增大看 bagging 误差趋稳。

```
import numpy as np
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
rng=np.random.default_rng(0)
x=np.sort(rng.uniform(0,1,200)); y=np.sin(6*x)+0.2*rng.standard_normal(200)
Xtr,Xte,ytr,yte=train_test_split(x[:,None],y,test_size=0.4,random_state=0)
tree=DecisionTreeRegressor(max_depth=8,random_state=0).fit(Xtr,ytr)
preds=np.zeros((100,len(Xte)))
for b in range(100):
    idx=rng.integers(0,len(Xtr),len(Xtr))
    preds[b]=DecisionTreeRegressor(max_depth=8,random_state=b).fit(Xtr[idx],ytr[idx]).predict(Xte)
print('单棵 MSE', round(mean_squared_error(yte,tree.predict(Xte)),4))
print('bagging MSE', round(mean_squared_error(yte,preds.mean(0)),4))
```

```
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif']=['SimHei']
steps=np.arange(1,101)
err=[mean_squared_error(yte,preds[:s].mean(0)) for s in steps]
plt.plot(steps,err)
plt.xlabel('B'); plt.ylabel('bagging测试MSE')
plt.title('Bagging 随 B 收敛')
plt.show()
```
