本笔记本逐步复现"维数灾维"：随着维数增加，局部邻域迅速变空。

## 1. 球与立方体的体积比

在单位超立方体中放置半径 $r=0.5$ 的超球，其体积占比随维数指数下降。

```
import math, numpy as np
def share(p): return math.pi**(p/2)/math.gamma(p/2+1)*(0.5**p)
print('p=10 占比 =', round(share(10), 5))
```

```
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
p = np.arange(1,21)
plt.plot(p, [share(x) for x in p], 'o')
plt.xlabel('维数 p'); plt.ylabel('占比')
plt.title('维数灾维')
plt.show()
```
