# 5.4 收缩方法: 岭与 Lasso（深讲）

比较 $\ell_2$（岭）与 $\ell_1$（Lasso）罚项对系数的影响。

## 1. 岭回归存在的闭式解

$$\hat\beta^{ridge}=(X^TX+\lambda I)^{-1}X^Ty$$

## 2. 系数路径与稀疏性

观察随 $\lambda$ 增大岭系数连续缩小；Lasso 使部分系数精确为零。

```
import numpy as np
from sklearn.datasets import make_regression
from sklearn.linear_model import Ridge
X,y=make_regression(n_samples=80,n_features=8,n_informative=5,noise=0.5,random_state=0)
al=np.logspace(-3,3,30)
C=np.array([Ridge(alpha=a).fit(X,y).coef_ for a in al])
print('岭系数矩阵形状', C.shape)
```

```
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif']=['Arial Unicode MS','PingFang SC','SimHei']
plt.plot(np.log10(al), C)
plt.xlabel('log10(lambda)'); plt.ylabel('系数')
plt.title('岭回归系数路径'); plt.show()
```

```
from sklearn.linear_model import Lasso
lasso=Lasso(alpha=0.5,max_iter=5000).fit(X,y)
print('Lasso 非零系数个数', int(np.sum(lasso.coef_!=0)))
```
