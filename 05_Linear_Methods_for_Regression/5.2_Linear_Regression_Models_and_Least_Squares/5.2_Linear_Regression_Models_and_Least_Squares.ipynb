## 2. 闭式求解

用 `np.linalg.solve` 求解线性方程组，避免直接求逆，数值更稳。

```
import numpy as np
from sklearn.datasets import load_diabetes
X, y = load_diabetes(return_X_y=True)
Xb = np.c_[np.ones(len(X)), X]
print('设计矩阵形状:', Xb.shape)
```

```
beta = np.linalg.solve(Xb.T @ Xb, Xb.T @ y)
rss = np.sum((y - Xb@beta)**2)
print('beta0 =', round(beta[0],3), ' RSS =', round(rss,2))
```

```
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif']=['Arial Unicode MS','PingFang SC','SimHei']
yhat = Xb@beta
plt.scatter(yhat, y-yhat, s=10)
plt.axhline(0,c='k',lw=0.8)
plt.xlabel('拟合值'); plt.ylabel('残差'); plt.title('残差-拟合图')
plt.show()
```
