## 2. 与 OLS 对比

在强相关问题下，PCR/PLS 以偏差换方差。

```
import numpy as np
from sklearn.datasets import make_regression
from sklearn.decomposition import PCA
from sklearn.linear_model import LinearRegression
from sklearn.cross_decomposition import PLSRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
X,y=make_regression(n_samples=120,n_features=10,noise=0.6,random_state=1)
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=0.3,random_state=0)
```

```
pca=PCA(n_components=4); Xp=pca.fit_transform(Xtr)
m=LinearRegression().fit(Xp,ytr)
print('PCR(4) 测试MSE =', round(mean_squared_error(yte, m.predict(pca.transform(Xte))),2))
```

```
pls=PLSRegression(n_components=4).fit(Xtr,ytr)
print('PLS(4) 测试MSE =', round(mean_squared_error(yte, pls.predict(Xte)),2))
```
