# 6.3 线性判别分析 (LDA)（深讲）

推导判别函数并在鸢尾花数据上可视化决策区域。

## 1. 判别函数

$$\delta_k(x)=x^T\Sigma^{-1}\mu_k-\tfrac12\mu_k^T\Sigma^{-1}\mu_k+\log\pi_k$$

## 2. 实例

使用鸢尾花前两个特征。

```
import numpy as np
from sklearn.datasets import load_iris
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
iris=load_iris(); X=iris.data[:,:2]; y=iris.target
lda=LinearDiscriminantAnalysis().fit(X,y)
print('训练准确率', round(lda.score(X,y),3))
```

```
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif']=['Arial Unicode MS','PingFang SC','SimHei']
gx,gy=np.meshgrid(np.linspace(X[:,0].min()-1,X[:,0].max()+1,300),np.linspace(X[:,1].min()-1,X[:,1].max()+1,300))
Z=lda.predict(np.c_[gx.ravel(),gy.ravel()]).reshape(gx.shape)
plt.contourf(gx,gy,Z,alpha=0.3,cmap='Pastel1')
plt.scatter(X[:,0],X[:,1],c=y,cmap='Set1',s=15)
plt.title('LDA 决策区域')
plt.show()
```
