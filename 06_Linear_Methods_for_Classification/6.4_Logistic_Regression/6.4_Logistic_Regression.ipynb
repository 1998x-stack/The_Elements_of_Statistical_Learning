## 2. 决策边界

在两类合成数据上可视化决策区域；注意边界仍是线性。

```
import numpy as np
from sklearn.datasets import make_blobs
from sklearn.linear_model import LogisticRegression
X,y=make_blobs(n_samples=300,centers=2,cluster_std=2.0,random_state=0)
clf=LogisticRegression().fit(X,y)
print('准确率', round(clf.score(X,y),3), '系数', np.round(clf.coef_[0],2))
```

```
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif']=['Arial Unicode MS','PingFang SC','SimHei']
gx,gy=np.meshgrid(np.linspace(X[:,0].min()-2,X[:,0].max()+2,300),np.linspace(X[:,1].min()-2,X[:,1].max()+2,300))
pp=clf.predict_proba(np.c_[gx.ravel(),gy.ravel()])[:,1].reshape(gx.shape)
plt.contourf(gx,gy,pp,alpha=0.5,cmap='RdBu')
plt.scatter(X[:,0],X[:,1],c=y,cmap='bwr',s=12)
plt.colorbar()
plt.title('逻辑回归概率')
plt.show()
```
