# 14.2 支持向量分类器（深讲）

最大间隔超平面 + 软间隔。

## 1. 模型

$$\min\tfrac12\|\beta\|^2+C\sum\xi_i$$

## 2. 边界

绘制线性软间隔决策面。

```
import numpy as np
from sklearn.svm import LinearSVC
from sklearn.datasets import make_blobs
X,y=make_blobs(n_samples=120,centers=2,cluster_std=1.5,random_state=0)
clf=LinearSVC(C=1.0).fit(X,y)
print('准确率', round(clf.score(X,y),3))
```

```
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif']=['SimHei']
gx,gy=np.meshgrid(np.linspace(X[:,0].min()-1,X[:,0].max()+1,250),np.linspace(X[:,1].min()-1,X[:,1].max()+1,250))
z=clf.predict(np.c_[gx.ravel(),gy.ravel()]).reshape(gx.shape)
plt.contourf(gx,gy,z,alpha=0.3,cmap='bwr')
plt.scatter(X[:,0],X[:,1],c=y,cmap='bwr',s=12)
plt.title('支持向量分类器')
plt.show()
```
