## 2. RBF

比较 RBF 与线性核的边界。

```
import numpy as np
from sklearn.svm import SVC
from sklearn.datasets import make_moons
X,y=make_moons(n_samples=250,noise=0.12,random_state=0)
svc=SVC(kernel='rbf',gamma=2).fit(X,y)
print('RBF 训练准确率', round(svc.score(X,y),3))
```

```
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif']=['SimHei']
fig,ax=plt.subplots(1,2,figsize=(10,4))
gx,gy=np.meshgrid(np.linspace(X[:,0].min()-.2,X[:,0].max()+.2,250),np.linspace(X[:,1].min()-.2,X[:,1].max()+.2,250))
for axx,kern in zip(ax,['linear','rbf']):
    m=SVC(kernel=kern).fit(X,y)
    z=m.predict(np.c_[gx.ravel(),gy.ravel()]).reshape(gx.shape)
    axx.contourf(gx,gy,z,alpha=0.3,cmap='bwr')
    axx.scatter(X[:,0],X[:,1],c=y,cmap='bwr',s=8)
    axx.set_title(kern)
plt.show()
```
