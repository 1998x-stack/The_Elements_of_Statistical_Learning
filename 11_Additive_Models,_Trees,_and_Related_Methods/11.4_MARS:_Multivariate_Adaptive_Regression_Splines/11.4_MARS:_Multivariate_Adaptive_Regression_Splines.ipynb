# 11.4 MARS（深讲）

用 hinge 基前向逐步拟合一维分段线性函数。

## 1. 候选基

$(x-\xi)_+ = \max(0, x-\xi)$。

## 2. 前向选择

每次加入使残差平方和最小的 hinge 再线性拟合。

```
import numpy as np
from sklearn.linear_model import LinearRegression
rng=np.random.default_rng(0)
x=np.sort(rng.uniform(0,1,200))
y=2+5*np.maximum(x-0.4,0)-3*np.maximum(x-0.7,0)+0.1*rng.standard_normal(200)
def hinge(a,k): return np.maximum(a-k,0)
K=[]
for _ in range(4):
    best=None
    for k in x:
        B=np.column_stack([np.ones_like(x)]+[hinge(x,kk) for kk in K+[k]])
        m=LinearRegression().fit(B,y)
        rss=np.sum((y-m.predict(B))**2)
        if best is None or rss<best[0]: best=(rss,k)
    K.append(best[1])
B=np.column_stack([np.ones_like(x)]+[hinge(x,k) for k in K])
m=LinearRegression().fit(B,y)
print('选中折点', np.round(K,3))
```

```
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif']=['SimHei']
xx=np.linspace(0,1,300)
Bx=np.column_stack([np.ones_like(xx)]+[hinge(xx,k) for k in K])
plt.scatter(x,y,s=6,alpha=0.5)
plt.plot(xx,m.predict(Bx),lw=2)
plt.title('MARS 分段线性拟合')
plt.show()
```
