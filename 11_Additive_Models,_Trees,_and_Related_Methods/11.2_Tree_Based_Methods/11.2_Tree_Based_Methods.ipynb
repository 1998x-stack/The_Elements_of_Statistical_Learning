# 11.2 树方法（深讲）

用递归二分构造决策树并可视化轴平行边界。

## 1. 划分

按特征阈值递归切分, 叶内给常数预测。

## 2. 边界

深度影响分段数目。

```
import numpy as np
from sklearn.datasets import make_circles
from sklearn.tree import DecisionTreeClassifier
X,y=make_circles(n_samples=300,noise=0.1,factor=0.5,random_state=0)
clf=DecisionTreeClassifier(max_depth=4,random_state=0).fit(X,y)
print('深度',clf.get_depth(),'叶',clf.get_n_leaves())
```

```
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif']=['SimHei']
gx,gy=np.meshgrid(np.linspace(X[:,0].min()-0.2,X[:,0].max()+0.2,200),np.linspace(X[:,1].min()-0.2,X[:,1].max()+0.2,200))
z=clf.predict(np.c_[gx.ravel(),gy.ravel()]).reshape(gx.shape)
plt.contourf(gx,gy,z,alpha=0.3,cmap='bwr')
plt.scatter(X[:,0],X[:,1],c=y,cmap='bwr',s=10)
plt.title('决策树边界')
plt.show()
```
