# 13.3 神经网络（深讲）

单隐层网络 f_k = σ0(β0k + Σ w_jk g_j(x))，g_j 是隐激活。

## 1. 决策面

在二维合成数据上用 MLP 拟合并可视化。

```
import numpy as np
from sklearn.datasets import make_moons
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import train_test_split
X,y=make_moons(n_samples=300,noise=0.15,random_state=0)
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=0.3,random_state=0)
mlp=MLPClassifier(hidden_layer_sizes=(10,),activation='tanh',max_iter=1000,random_state=0).fit(Xtr,ytr)
print('测试准确率', round(mlp.score(Xte,yte),3))
```

```
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif']=['SimHei']
gx,gy=np.meshgrid(np.linspace(X[:,0].min()-.2,X[:,0].max()+.2,250),np.linspace(X[:,1].min()-.2,X[:,1].max()+.2,250))
z=mlp.predict(np.c_[gx.ravel(),gy.ravel()]).reshape(gx.shape)
plt.contourf(gx,gy,z,alpha=0.3,cmap='bwr')
plt.scatter(X[:,0],X[:,1],c=y,cmap='bwr',s=12)
plt.title('单隐层 NN 决策面')
plt.show()
```
