# 12.10 梯度提升（深讲）

在函数空间做梯度下降：每轮把负梯度（残差）拟合到小树上再累加。

## 1. 负梯度

平方损失下即残差 $y - f_{m-1}(x)$。

## 2. 收缩

以学习率 $\eta$ 缩放每步。

```
import numpy as np
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
rng=np.random.default_rng(0)
x=np.sort(rng.uniform(0,1,200)); y=np.sin(6*x)+0.2*rng.standard_normal(200)
Xtr,Xte,ytr,yte=train_test_split(x[:,None],y,test_size=0.4,random_state=0)
eta=0.1; Ftr=np.full_like(ytr,ytr.mean()); Fte=np.full_like(yte,ytr.mean())
for m in range(150):
    t=DecisionTreeRegressor(max_depth=3,random_state=m).fit(Xtr,ytr-Ftr)
    Ftr+=eta*t.predict(Xtr); Fte+=eta*t.predict(Xte)
print('GBM(150,深度3,eta=0.1) 测试MSE', round(mean_squared_error(yte,Fte),4))
```
