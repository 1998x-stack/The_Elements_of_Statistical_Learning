# 12.4 指数损失与 AdaBoost（深讲）

指数损失 $L(y,f)=e^{-y f(x)}$ 是 0-1 损失的光滑可微上界。

## 1. 权重更新

$$w_i \leftarrow w_i e^{-\alpha_m y_i G_m(x_i)}$$

```
import numpy as np
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import make_moons
X,y=make_moons(n_samples=200,noise=0.2,random_state=0); y=np.where(y==1,1,-1)
w=np.ones(len(X))/len(X); alphas=[]; ps=[]
for m in range(40):
    t=DecisionTreeClassifier(max_depth=1,random_state=m).fit(X,y,sample_weight=w)
    p=t.predict(X); er=np.mean(w*(p!=y))/np.sum(w)
    a=0.5*np.log((1-er)/(er+1e-12))
    w=w*np.exp(-a*y*p); w/=w.sum(); alphas.append(a); ps.append(p)
F=sum(a*p for a,p in zip(alphas,ps))
print('集成0-1误差', np.mean(np.sign(F)!=y))
```
