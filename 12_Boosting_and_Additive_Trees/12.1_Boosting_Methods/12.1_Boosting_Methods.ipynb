# 12.1 Boosting（深讲）

手写 AdaBoost：每轮对弱学习器重新加权，组合成强学习器。

## 1. 权重更新

$$w_i \leftarrow w_i e^{\alpha_m\mathbf 1_{[G_m(x_i)\ne y_i]}}$$

```
import numpy as np
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import make_moons
X,y=make_moons(n_samples=200,noise=0.2,random_state=0); y=np.where(y==1,1,-1)
w=np.ones(len(X))/len(X); al=np.array([]); ts=[]
for m in range(30):
    t=DecisionTreeClassifier(max_depth=1,random_state=m).fit(X,y,sample_weight=w)
    pred=t.predict(X); err=np.sum(w*(pred!=y))/w.sum()
    a=0.5*np.log((1-err)/(err+1e-12))
    w=w*np.exp(-a*y*pred); w=w/w.sum()
    ts.append(t); alphas=np.append(alphas,a)
F=sum(ai*t.predict(X) for ai,t in zip(alphas,ts))
print('训练误差', np.mean(np.sign(F)!=y))
```
